```python
from pathlib import Path
from collections import Counter

data_path = Path(r"C:\GymProject\data")

if not data_path.exists():
    print("المسار غير موجود، عدّل data_path")
else:
    print("المجلدات الرئيسية:")
    for item in sorted(data_path.iterdir()):
        print(item.name)

    extensions = {".mp4", ".avi", ".mov", ".mkv"}
    videos = [
        p for p in data_path.rglob("*")
        if p.is_file() and p.suffix.lower() in extensions
    ]

    print("\nعدد الفيديوهات:", len(videos))
    print("\nأمثلة لمسارات الفيديوهات:")
    for video in videos[:15]:
        print(video.relative_to(data_path))

    print("\nأنواع الملفات:")
    print(Counter(p.suffix.lower() for p in videos))
```

```python
from collections import Counter
import re

root = data_path / "Dataset Exercise Quality-wise"

if not root.is_dir():
    raise FileNotFoundError(f"المجلد غير موجود: {root}")

counts = Counter()
subjects = set()

for video in root.rglob("*.mp4"):
    parts = video.relative_to(root).parts

    if len(parts) >= 4:
        exercise, quality, view = parts[:3]
        counts[(exercise, quality, view)] += 1

    match = re.search(r"subject_(\d+)", video.name, re.IGNORECASE)
    if match:
        subjects.add(match.group(1))

print("التمارين الموجودة:")
print(sorted({key[0] for key in counts}))

print("\nعدد الأشخاص:", len(subjects))

print("\nExercise | Quality | View | Videos")
print("-" * 65)
for (exercise, quality, view), count in sorted(counts.items()):
    print(f"{exercise:15} | {quality:7} | {view:10} | {count}")
```
